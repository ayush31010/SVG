# SVG Graph-MoE: matched set grounding
Select a T4 GPU and run all. Trains the GNN and no-edge control on identical source-disjoint synthetic data, then evaluates the frozen natural subset. Results remain exploratory.


```python
import os, sys, subprocess, io, zipfile, urllib.request
from pathlib import Path
import torch
assert torch.cuda.is_available(), 'Select a GPU runtime'
print(torch.cuda.get_device_name(0), torch.__version__)
REPO_REF = 'agent/svg-colab-set-v5'
url = f'https://codeload.github.com/ayushdebnath012/SVG/zip/refs/heads/{REPO_REF}'
workspace = Path('/content/svg-cloud') if Path('/content').exists() else Path('/kaggle/working/svg-cloud')
workspace.mkdir(parents=True, exist_ok=True)
with urllib.request.urlopen(url, timeout=120) as response:
    archive = zipfile.ZipFile(io.BytesIO(response.read()))
    root_name = archive.namelist()[0].split('/')[0]
    archive.extractall(workspace)
repo = workspace / root_name
os.chdir(repo)
sys.path.insert(0, str(repo))
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'numpy', 'Pillow', 'cairosvg', 'svgelements', 'pyarrow'], check=True)
print('READY', repo)
```
