# Frozen new-collection group candidate experiment
100 reviewed instructions, 50 new SVG sources, three arms. Source and protocol are pinned before scoring.


```python
import os, sys, io, zipfile, urllib.request, subprocess
from pathlib import Path
import torch
assert torch.cuda.is_available(), 'Choose a T4 GPU runtime'
print('GPU', torch.cuda.get_device_name(0), torch.__version__, flush=True)
SOURCE_COMMIT = '92c8a985c59bf05133b54233cb2c55b6527613c0'
workspace = Path('/content/svg-group') if Path('/content').exists() else Path('/kaggle/working/svg-group')
workspace.mkdir(parents=True, exist_ok=True)
with urllib.request.urlopen(f'https://codeload.github.com/ayushdebnath012/SVG/zip/{SOURCE_COMMIT}', timeout=120) as response:
    archive = zipfile.ZipFile(io.BytesIO(response.read()))
    root_name = archive.namelist()[0].split('/')[0]
    archive.extractall(workspace)
repo = workspace / root_name
os.chdir(repo)
sys.path.insert(0, str(repo))
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'numpy', 'Pillow', 'svgelements'], check=True)
print('READY', repo, flush=True)
```
