# SVG Graph-MoE: matched set grounding
Select a T4 GPU and run all. Trains the GNN and no-edge control on identical source-disjoint synthetic data, then evaluates the frozen natural subset. Results remain exploratory.
Uses a pinned GitHub source revision. The final cell evaluates learned-count visual fusion and downloads metrics plus both checkpoints; retain the ZIP before the temporary runtime expires. Kaggle also requires Internet access enabled.


```python
import os, sys, subprocess, io, zipfile, urllib.request
from pathlib import Path
import torch
assert torch.cuda.is_available(), 'Select a GPU runtime'
print(torch.cuda.get_device_name(0), torch.__version__)
REPO_REF = '470ca986045d15dc3aa3f472c99f7285174dcb3a'  # immutable experiment source
url = f'https://codeload.github.com/ayushdebnath012/SVG/zip/{REPO_REF}'
workspace = Path('/content/svg-cloud') if Path('/content').exists() else Path('/kaggle/working/svg-cloud')
workspace.mkdir(parents=True, exist_ok=True)
with urllib.request.urlopen(url, timeout=120) as response:
    archive = zipfile.ZipFile(io.BytesIO(response.read()))
    root_name = archive.namelist()[0].split('/')[0]
    archive.extractall(workspace)
repo = workspace / root_name
os.chdir(repo)
sys.path.insert(0, str(repo))
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'numpy', 'Pillow', 'cairosvg', 'svgelements', 'pyarrow'], check=True)
print('READY', repo)
```

```python
from scripts.analyze_cloud_cardinality_fusion import run as analyze_fusion
from pathlib import Path
import zipfile
output = Path('runs/colab-set-v4').resolve()
fusion = analyze_fusion(output, output / 'vector-edits-test.parquet')
archive = output.with_suffix('.zip')
# Refresh the original archive so the additional analysis is preserved too.
with zipfile.ZipFile(archive, 'w', zipfile.ZIP_DEFLATED) as bundle:
    for path in sorted(output.rglob('*')):
        if path.is_file() and path.name != 'vector-edits-test.parquet':
            bundle.write(path, path.relative_to(output))
print('Download results and both checkpoints:', archive)
try:
    from google.colab import files
    files.download(str(archive))
except ImportError:
    from IPython.display import FileLink, display
    display(FileLink(str(archive)))
```
