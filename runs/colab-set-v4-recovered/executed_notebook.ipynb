```python
import os, sys, subprocess, io, zipfile, urllib.request
from pathlib import Path
import torch
assert torch.cuda.is_available(), 'Select a GPU runtime'
print('GPU',torch.cuda.get_device_name(0),torch.__version__,flush=True)
url = 'https://codeload.github.com/ayushdebnath012/SVG/zip/075687c'
workspace = Path('/content/svg-cloud')
workspace.mkdir(parents=True,exist_ok=True)
with urllib.request.urlopen(url,timeout=120) as response:
    archive = zipfile.ZipFile(io.BytesIO(response.read()))
    root_name = archive.namelist()[0].split('/')[0]
    archive.extractall(workspace)
repo = workspace / root_name
os.chdir(repo)
sys.path.insert(0,str(repo))
r = subprocess.run([sys.executable,'-m','pip','install','-q','numpy','Pillow','cairosvg','svgelements','pyarrow'],capture_output=True,text=True)
print(r.stdout,r.stderr)
r.check_returncode()
print('READY',repo,flush=True)
```
